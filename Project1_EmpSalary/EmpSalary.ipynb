# Train and Test

```python
df_model = df[
    (df["salary_midpoint_lpa"] > 0)&
    (df['salary_midpoint_lpa'].notna())&
    (df['salary_midpoint_lpa'] < np.inf)
].copy()

print("Original result: ", len(df))
print("Usable rows:", len(df_model))
```

```python
features = [
    "experience_min_yrs",
    "experience_max_yrs",
    "company_rating",
    "skills_count"
]

target = "salary_midpoint_lpa"

X = df_model[features]
y = df_model[target]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
print("Training salary mean:", y_train.mean())
print("Testing salary mean:", y_test.mean())

print("Missing features:", X_train.isnull().sum().sum())
print("Missing target:", y_train.isnull().sum())
```

# Linear Regression Model Train

```python
from sklearn.linear_model import LinearRegression

lm = LinearRegression()
lm.fit(X_train, y_train)

print("Model trained successfully!")
print("Intercept:", lm.intercept_)
print("Coefficients:", lm.coef_)
```

```python
y_pred = lm.predict(X_test)

print("Actual salaries:")
print(y_test.head().to_list())

print("Predicted salaries:")
print(y_pred[:5])
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
import numpy as np

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

r = np.corrcoef(y_test, y_pred)[0, 1]

print(f"MAE:  {mae:.3f} LPA")
print(f"MSE:  {mse:.3f} (LPA)^2")
print(f"RMSE: {rmse:.3f} LPA")
print(f"R²:   {r2:.3f}")
print(f"R:    {r:.3f}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 6))
plt.scatter(y_test, y_pred, alpha=0.4)

low = min(y_test.min(), y_pred.min())
high = max(y_test.max(), y_pred.max())

plt.plot([low, high], [low, high], "r--")
plt.xlabel("Actual Salary (LPA)")
plt.ylabel("Predicted Salary (LPA)")
plt.title("Actual vs Predicted Salary")
plt.show()
```

```python
residuals = y_test - y_pred

plt.figure(figsize=(7, 5))
plt.scatter(y_pred, residuals, alpha=0.4)
plt.axhline(0, color="red", linestyle="--")
plt.xlabel("Predicted Salary (LPA)")
plt.ylabel("Residual (LPA)")
plt.title("Residual Plot")
plt.show()
```

```python
from sklearn.dummy import DummyRegressor

baseline = DummyRegressor(strategy="mean")
baseline.fit(X_train, y_train)

baseline_pred = baseline.predict(X_test)

print("Baseline MAE:",
      mean_absolute_error(y_test, baseline_pred))
print("Linear Regression MAE:", mae)
```
